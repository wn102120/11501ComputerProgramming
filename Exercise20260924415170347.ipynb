30
 （或 32 位元系統下的 2 
15

儲存空間擴充機制：當數值變大超出當前陣列容量時，CPython 會動態重新分配（reallocate）更大的記憶體空間來儲存更多的 digit 區塊。

2. 實數 (float)
是否為內建型別：是。

CPython 物件結構名稱：PyFloatObject。   
PNG

資料儲存方式：內部直接包裝一個 C 語言的 double 變數，依據 IEEE 754 標準以 64 位元（1 位元符號 + 11 位元指數 + 52 位元小數）儲存。   
PNG

固定大小限制：有固定大小限制。通常最大值約為 1.79×10 
308
 ，最小正值約為 2.22×10 
−308
 。

儲存空間擴充機制：無法擴充。每個 float 物件佔用的記憶體大小固定，若數值超出上限會引發 OverflowError，若低於下限則會下溢為 0.0。

3. 複數 (complex)
是否為內建型別：是。

CPython 物件結構名稱：PyComplexObject。   
PNG

資料儲存方式：內部包裝一個 Py_complex 結構，由兩個 C 語言的 double 變數組成，分別儲存實部（real）與虛部（imag）。   
PNG

固定大小限制：有固定大小限制（實部與虛部各自受限於 IEEE 754 double 的範圍）。

儲存空間擴充機制：無法擴充。與 float 相同，實部與虛部的記憶體大小均固定為 64 位元。

4. 集合 (set)
是否為內建型別：是。

CPython 物件結構名稱：PySetObject。   
PNG

資料儲存方式：基於雜湊表（Hash Table）儲存，內部維護一個指向 setentry 結構陣列的指標，每個條目包含 key 的指標與其雜湊值（hash value）。   
PNG

固定大小限制：無固定元素數量限制，僅受限於可用記憶體。

儲存空間擴充機制：當集合填滿程度達到一定比例（通常是表格容量的 2/3）時，CPython 會觸發重構（Resize），分配一個約為原本 2 到 4 倍大的新雜湊表，並將所有元素重新雜湊（Rehash）放置於新表中。

5. 命題真假 (bool)
是否為內建型別：是。

CPython 物件結構名稱：PyBoolObject（底層直接繼承自 PyLongObject）。   
PNG

資料儲存方式：繼承自整數，本質上以 int 的方式儲存數值 0（False）或 1（True）。   
PNG

固定大小限制：有固定限制。全域僅有兩個預先建立好的單例物件（Py_True 與 Py_False）。   
PNG

儲存空間擴充機制：不需要擴充。因為布林值僅有兩種狀態，全域共用單例，不會新增額外資料規模。

6. 串列 (list)
是否為內建型別：是。

CPython 物件結構名稱：PyListObject。   
PNG

資料儲存方式：以連續記憶體空間的「動態陣列（Dynamic Array）」實作，陣列中儲存的是指向各個 Python 物件的指標（PyObject*）。   
PNG

固定大小限制：無固定長度限制，僅受限於系統記憶體（或 64 位元系統下索引的最大上限 2 
63
 −1）。

儲存空間擴充機制：當執行 append 或 extend 且容量不足時，CPython 會進行 Over-allocation（過度分配），自動分配比目前需求更大的連續記憶體空間（擴容係數約為 1.125∼1.25 加常數），以減少頻繁重新分配記憶體的時間開銷。

## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？
Python 的整數型別採用了「任意精度（Arbitrary Precision）」機制。當數值超出傳統固定位元的範圍時，Python 會自動分配更多記憶體空間，將數字拆解成多個區塊進行儲存與計算，因此不受傳統硬體暫存器大小的限制。
2. Python 的 `int` 是否具有固定的位元數限制？
沒有
3. 什麼是 Arbitrary Precision（任意精度）？
Arbitrary Precision 指的是數字的精度或大小不受硬體固定位元數（如 32 或 64 位元）限制，而是根據數字大小動態分配記憶體。只要系統記憶體足夠，數字就可以擴展到任意長度。
4. CPython 使用什麼物件來實作 `int`？
使用 PyLongObject 結構來實作
5. Python 整數大小真正的限制是什麼？
真正的限制是電腦的可用記憶體（RAM）以及計算處理時間

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？
Python 的 int 採用了任意精度（Arbitrary Precision）的實作方式。在 CPython 底層，整數並非直接映射到 CPU 固定的 32 位元或 64 位元暫存器，而是封裝在 PyLongObject 結構中，以動態變長陣列來儲存數值區塊。當數值變大時，Python 會自動向作業系統申請更多記憶體來擴充儲存空間，因此其大小僅受限於電腦的可用記憶體（RAM）。
2. 為什麼 `float` 卻存在範圍限制？
由於指數位的長度固定為 11 位元，其能表示的最大數值極限被嚴格鎖定在大約 $1.79 \times 10^{308}$，超過此範圍就會引發數值溢位
3. 為什麼 `float` 可能產生誤差？
二進位與十進位的轉換問題：
電腦內部僅能以二進位（$2^n$ 的組合）儲存資料。十進位中的某些有限小數（如 0.1 或 0.2），轉為二進位後會變成無限循環小數。

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。
Python 的整數（Arbitrary Precision / 正確性優先）：
核心哲學是「不惜代價確保絕對精確」。系統會根據資料大小動態分配記憶體，不預先設定數值邊界，確保計算過程中完全不產生溢位或精確度損失。

AI 模型的 FP16 / BF16（Limited Precision / 效率優先）：
核心哲學是「在可接受的誤差範圍內極大化計算效率」。將資料鎖定在固定的 16 位元空間，犧牲部分數值細節（尾數精度），換取記憶體佔用減半與硬體平行計算速度的暴增。
2. 各自的優點與缺點。
Python整數 
 優點 : 零誤差：無溢位風險，結果百分之百精確。
 缺點 : 效能低落：動態分配記憶體開銷大。
FP16 / BF16
 優點 : 極高吞吐量：可完美對應 GPU Tensor Cores 等硬體加速。
 缺點 : 存在誤差：無法精確表示微小細節。 
3. 為什麼大型語言模型通常選擇較低精度的表示方式？
神經網絡具有極高的「抗噪性（Fault Tolerance）」：
大語言模型（如 Llama, GPT 等）包含數百億到數千億個參數。模型輸出的是概率分佈，微小的權重精確度偏差並不會改變整體語義或最終的 Token 預測結果。

解決「顯存牆（Memory Wall）」與傳輸瓶頸：
模型訓練與推論的主要瓶頸往往在於顯存頻寬（Memory Bandwidth）。採用 16 位元格式可讓顯存佔用量減半，提升 GPU 傳輸資料與快取的效率。

硬體加速器的原生支援：
現代 GPU（如 NVIDIA A100, H100 等）內建的 Tensor Cores 專門針對 FP16/BF16 矩陣乘法進行硬體優化，計算速度比傳統 FP32/FP64 快上數倍。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？
顯存爆炸、算力巨大浪費與訓練時間激增
