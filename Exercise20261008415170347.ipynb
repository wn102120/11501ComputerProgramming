習題一：使用random模組，產生出一個介於0到1之間的小數$r$，使用while迴圈，計算出$r$的前二十位二進位表示$r = 0.b_1b_2b_3\cdots {b_{20}}_2$。輸出格式如附圖Exercise20261015.png

import random

# 1. 產生 0 到 1 之間的隨機小數 r
r = random.random()
print(f"假如 r = {r}")

# 備份原始的 r 以便運算
temp_r = r
binary_bits = ""
count = 0

# 2. 使用 while 迴圈計算前 20 位二進位小數
while count < 20:
    temp_r *= 2
    bit = int(temp_r)          # 取出整數部分 (0 或 1)
    binary_bits += str(bit)     # 串接位元
    temp_r -= bit               # 扣除整數部分，保留小數
    count += 1

# 3. 按照題目要求的格式輸出 (下標 2 使用 Unicode '\u2082')
print(f"r \u2248 0.{binary_bits}\u2082")

Python 的 `int` 型別採用 Arbitrary-Precision Integer（任意精度整數）設計，能夠精確表示任意大小的整數，其實作方式可參考 CPython 的 `PyLongObject`。與 C/C++ 的 `int` 或 `long long` 不同，Python 的整數不受固定 32 位元或64 位元限制，而是會依據數值大小自動配置更多記憶體，因此只要記憶體足夠，便能表示極大的整數。例如：
```python
n = 10**68
print(n)
```
上述程式可正常執行，表示 Python 能夠精確儲存$10^{68}$，即中文數字中的「無量大數」。


可利用下列程式觀察不同大小整數所需的位元數（bit）與記憶體空間：

```python
import sys
n = 10**68
print(f"bit_length = {n.bit_length()}")
print(f"memory = {sys.getsizeof(n)} bytes")
```

習題二：輸入一個正整數 \(n\)，其中$1 \le n \le 10^{68}$，請將 \(n\) 轉換為中文大寫數字表示法。需支援下列數字單位：萬、億、兆、京、垓、秭、穰、溝、澗、正、載、極、恆河沙、阿僧祇、那由他、不可思議、無量大數。比如輸入31415926，輸出參仟壹佰肆拾壹萬伍仟玖佰貳拾陸。

提示：思考利用字典，建立「數字」與「中文大寫數字」的對應關係。

def int_to_chinese(n: int) -> str:
    # 中文大寫數字與小單位
    num_chars = ["零", "壹", "貳", "參", "肆", "伍", "陸", "柒", "捌", "玖"]
    sub_units = ["", "拾", "佰", "仟"]
    
    # 大單位列表（每 4 位為一級）
    big_units = [
        "", "萬", "億", "兆", "京", "垓", "秭", "穣", "溝", "澗", 
        "正", "載", "極", "恆河沙", "阿僧祇", "那由他", "不可思議", "無量大數"
    ]
    
    s = str(n)
    length = len(s)
    
    # 將字串從右到左每 4 位切成一組
    groups = []
    while s:
        groups.append(s[-4:])
        s = s[:-4]
        
    res = []
    zero_flag = False  # 用於處理連續零的補零邏輯
    
    for i, group in enumerate(groups):
        group_int = int(group)
        if group_int == 0:
            # 若整組皆為 0，且不是最低組，則可能需要補一個「零」
            if i > 0 and not zero_flag:
                res.append("零")
                zero_flag = True
            continue
            
        zero_flag = False
        group_str = ""
        g_len = len(group)
        
        for j, char in enumerate(group):
            digit = int(char)
            pos = g_len - 1 - j  # 當前位數在組內的權重 (0:個, 1:拾, 2:佰, 3:仟)
            
            if digit != 0:
                # 若前面有略過的零且不是組首，補一個「零」
                if zero_flag:
                    group_str += "零"
                    zero_flag = False
                group_str += num_chars[digit] + sub_units[pos]
            else:
                zero_flag = True
                
        # 加上大單位（萬、億、兆...）
        group_str += big_units[i]
        res.append(group_str)
        
    # 由於是從低位到高位處理，反轉回來並組合
    return "".join(reversed(res))

# 測試範例
n = 31415926
print(f"輸入：{n}")
print(f"輸出：{int_to_chinese(n)}")
