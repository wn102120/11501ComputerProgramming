1. 0.1的二進位表示法為何? 使用markdown語法，列出計算過程和理由，並描述看到的現象。

十進位的0.1轉換為二進位時，採用乘2取整數法：
0.1 x 2 = 0.2，取整數 0
0.2 x 2 = 0.4，取整數 0
0.4 x 2 = 0.8，取整數 0
0.8 x 2 = 1.6，取整數 1，餘 0.6
0.6 x 2 = 1.2，取整數 1，餘 0.2
0.2 x 2 = 0.4，取整數 0，開始無限循環
二進位表示法：0.0001100110011..
現象：十進位中有限的小數0.1，在二進位表示法中會變成無限循環小數。由於電腦儲存有限，無法無限紀錄，因此會產生誤差。

3. Python如何知道浮點數的精度？請問執行以下程式碼:
```python
import sys
print(sys.float_info)
```
並說明輸出結果的意思。

輸出結果如下：
sys.float_info(max=1.7976931348623157e+308, max_exp=1024, max_10_exp=308, min=2.2250738585072014e-308, min_exp=-1021, min_10_exp=-307, dig=15, mant_dig=53, epsilon=2.220446049250313e-16, radix=2, rounds=1)
輸出結果代表的意思：
dig = 15：十進位浮點數可確保可靠精確的位數為15位。
mant_dig = 53：二進位尾數的長度為53個位元。
epsilon = 2.220446049250313e-16：1.0與比1.0大的下一個最小浮點數之間的差值，代表精度的極限。
max / min：浮點數能表示的最大值與最小正數。
